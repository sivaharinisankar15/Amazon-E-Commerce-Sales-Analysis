```python
# Task 2: Data Cleaning & Preprocessing
import pandas as pd
df = pd.read_csv("archive/Amazon Sale Report.csv")

print("Missing Values Before Cleaning:")
print(df[["Amount", "Qty", "Courier Status"]].isnull().sum())

df["Amount"] = df["Amount"].fillna(df["Amount"].median())
df["Qty"] = df["Qty"].fillna(df["Qty"].median())
df["Courier Status"] = df["Courier Status"].fillna("Unknown")

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
# df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
# df["Date"].dtype

df = df.drop_duplicates()

df["ship-state"] = df["ship-state"].str.strip().str.upper()

print("\nMissing Values After Cleaning:")
print(df[["Amount", "Qty", "Courier Status"]].isnull().sum())

print("\nData Types After Cleaning:")
print(df.dtypes)

print("\nDataset Shape After Cleaning:")
print(df.shape)

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())


# 7. Export cleaned dataset

# Export cleaned dataset
df.to_csv("archive/cleaned_amazon_sales.csv", index=False)


print("\nCleaned CSV file created successfully.")
```
